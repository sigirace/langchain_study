### 11.1 Audio Extraction

```python
import subprocess

def extract_audio_from_video(video_path, audio_path):
    command = ["ffmpeg", "-i", video_path, "-vn", audio_path]
    subprocess.run(command)

extract_audio_from_video("./files/whisper_video.mp4", "./files/whisper_audio.mp3")
```

### 11.2 Cutting The Audio

```python
from pydub import AudioSegment

track = AudioSegment.from_mp3("./files/whisper_audio.mp3")

ten_minutes = 10 * 60 * 1000
```

```python
import math

chunks = math.ceil(len(track) / ten_minutes)

for i in range(chunks):
    start_time = i * ten_minutes
    end_time = (i + 1) * ten_minutes

    chunk = track[start_time:end_time]

    chunk.export(f"./files/chunks/chunk_{i}.mp3", format="mp3")
```

### 11.3 Whisper Transcript

```python
def cut_audio_in_chunks(audio_path, chunk_size, chunks_folder):
    track = AudioSegment.from_mp3(audio_path)
    chunk_len = chunk_size * 60 * 1000
    chunks = math.ceil(len(track) / chunk_len)
    for i in range(chunks):
        start_time = i * chunk_len
        end_time = (i + 1) * chunk_len

        chunk = track[start_time:end_time]

        chunk.export(f"./{chunks_folder}/chunk_{i}.mp3", format="mp3")
```

```python
import openai

transcript = openai.Audio.transcribe("whisper-1", open("./files/chunks/chunk_0.mp3", "rb"))
```

```python
def decode_unicode_escape(encoded_str):
    """
    유니코드 이스케이프 문자열을 디코딩하여 일반 텍스트로 변환하는 함수.

    :param encoded_str: 유니코드 이스케이프 문자열
    :return: 디코딩된 일반 텍스트
    """
    try:
        # 먼저 unicode_escape로 디코딩
        intermediate_str = encoded_str.encode('utf-8').decode('unicode_escape')
        # 다시 UTF-8로 디코딩
        decoded_str = intermediate_str.encode('latin1').decode('utf-8')
        return decoded_str
    except Exception as e:
        return f"디코딩 중 오류 발생: {e}"

decoded_script = decode_unicode_escape(transcript.text)
print(decoded_script)
```

```python
from typing import final
import glob

def transcribe_chunks(chunk_folder, destination):
    files = glob.glob(f"{chunk_folder}/*.mp3")
    final_transcript = ""
    for file in files:
        with open(file, "rb") as audio_file:
            transcript = openai.Audio.transcribe(
                "whisper-1",
                audio_file,
            )
            final_transcript += decode_unicode_escape(transcript["text"])
    with open(destination, "w") as file:
        file.write(final_transcript)

transcribe_chunks("./files/chunks", "./files/transcript.txt")
```
