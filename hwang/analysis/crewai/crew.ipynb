### 2. Agents : 작업을 수행하고, 결정을 내리고, 다른 agent들과 통신할 수 있는 자율 단위
- 특정 스킬로 특정 업무를 수행하는 팀의 구성원
- 작업을 다른 agent에게 위임이 가능하며, 작업이 완료되었다고 말할 수 있고, 다른 agent들과 통신이 가능, agent에게 도구를 줄수도 있음 
- agent는 구체적이어야 함 (LLM, GPT에게 많은 일을 하라하면 잘 못함)
- Agent를 만들기 위해선 role을 지정해야 함 (crew안에서 agent의 역할을 정해주는 것)
- 목표와 뒷 배경이 무엇인지도 알려줘야 함. (롤플레잉을 해줘야 더 잘 작동함. ex.너는 요리사야 이런 것)
